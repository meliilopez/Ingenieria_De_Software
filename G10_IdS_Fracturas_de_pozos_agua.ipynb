### 2.5.2 Análisis estadístico descriptivo

Se realiza un análisis estadístico descriptivo de las variables numéricas, calculando:

* Media
* Mediana
* Desvío estándar
* Valor mínimo
* Valor máximo
* Cuartiles (Q1, Q2, Q3)

Este análisis permite entender la dispersión de los datos, detectar posibles outliers y analizar si las variables presentan comportamientos muy diferentes entre sí.

````python
# @title Estadística descriptiva de las variables numéricas
excluir = ["id_base_fractura_adjiv", "idpozo", "anio_if", "mes_if", "anio_ff", "mes_ff", "anio_carga", "mes_carga", "mes", "anio"]

num_cols = [c for c in columnas_originales if pd.api.types.is_numeric_dtype(df_datos_energia[c]) and c not in excluir]
# Se suma la duración de la operación, ya expresada en días (numérica)
num_cols.append("duracion_fractura_dias")


def estadisticos(s):
	s = s.dropna()
	q1, q2, q3 = s.quantile([0.25, 0.50, 0.75])
	iqr = q3 - q1
	media, desvio = s.mean(), s.std()
	lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
	return pd.Series(
		{
			"conteo": s.count(),
			"media": media,
			"varianza": s.var(),
			"desvio": desvio,
			"moda": s.mode().iloc[0],
			"minimo": s.min(),
			"Q1": q1,
			"mediana (Q2)": q2,
			"Q3": q3,
			"maximo": s.max(),
			"rango": s.max() - s.min(),
			"lim_inf": lim_inf,
			"lim_sup": lim_sup,
			"CV (%)": desvio / media * 100 if media != 0 else np.nan,
		}
	)


resumen = df_datos_energia[num_cols].apply(estadisticos).T.round(2)

with pd.option_context("display.max_columns", None, "display.width", 250, "display.float_format", lambda x: f"{x:,.2f}"):
	display(resumen)
````

#### Prueba de hipótesis de normalidad

Antes de detectar outliers es necesario determinar si cada variable sigue una distribución normal, ya que influye en el método a utilizar.


Para cada variable numérica se plantea la siguiente prueba de hipótesis:

***H0 : la variable se distribuye de forma normal***

***Ha : La variable no se distribuye de forma normal***

*Nivel de significancia: 𝛼 = 0.05*

*Test utilizado: Kolmogorov-Smirnov / Lilliefors*


````python
# @title Test de normalidad por variable
df_dist = df_datos_energia[num_cols].copy()
vars_sin_cero = ["agua_inyectada_m3", "presion_maxima_psi", "potencia_equipos_fractura_hp"]
df_dist[vars_sin_cero] = df_dist[vars_sin_cero].replace(0, np.nan)


def test_normalidad(s, alfa=0.05):
	s = s.dropna()

	test = "Kolmogorov-Smirnov (Lilliefors)"
	res = stats.goodness_of_fit(stats.norm, s, statistic="ks", n_mc_samples=999, random_state=42)
	estadistico, p_value = res.statistic, res.pvalue

	return pd.Series(
		{
			"estadístico": round(estadistico, 4),
			"p-valor": round(p_value, 4),
			"asimetría": round(s.skew(), 2),
			"curtosis": round(s.kurt(), 2),
			"decisión (α = 0,05)": "Se rechaza H0" if p_value < alfa else "No se rechaza H0",
		}
	)


resultados_normalidad = df_dist.apply(test_normalidad).T
display(resultados_normalidad)
````

Para las 9 variables analizadas el p-value es menor que α = 0.05, por lo que **se rechaza H₀ en todos los casos**. Con un 95 % de confianza, se afirma que ninguna variable sigue una distribución normal. Se utiliza el **método no paramétrico del rango intercuartílico (IQR)** para el cálculo de outliers.

Además, se observa:
* Asimetría positiva fuerte y curtosis muy altas en `arena_bombeada_importada_tn`, `presion_maxima_psi`, `duracion_fractura_dias` y `co2_inyectado_m3`. Son colas largas a la derecha con valores extremos.

### 2.5.5 Detección de outliers

Como ninguna variable sigue una distribución normal, los outliers se detectan mediante el método de Tukey. A diferencia del z-score, este método no depende de la media ni del desvío estándar, que se ven inflados por los propios valores extremos, sino de los cuartiles, que son robustos frente a la asimetría.

Consideraciones:

* Los ceros de agua, presión y potencia se tratan como faltantes, por lo que no se evalúan como outliers.
* En `co2_inyectado_m3` y `arena_bombeada_importada_tn` la mayoría de los registros vale 0 porque no se usó ese insumo. Si se los incluye, Q1 y Q3 dan 0 y cualquier valor positivo sería outlier, por eso se evalúan sólo los registros que sí usaron el insumo.
* Un valor atípico no es necesariamente un error: puede ser una operación real fuera de lo común. Por eso, después de contarlos, se inspeccionan los casos extremos para clasificarlos.

````python
# @title Conteo de outliers por el método IQR
def outliers_iqr(s):
	s = s.dropna()
	q1, q3 = s.quantile([0.25, 0.75])
	iqr = q3 - q1
	lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
	mask = (s < lim_inf) | (s > lim_sup)
	return pd.Series(
		{
			"n": len(s),
			"Q1": q1,
			"Q3": q3,
			"IQR": iqr,
			"lim_inf": lim_inf,
			"lim_sup": lim_sup,
			"por debajo": (s < lim_inf).sum(),
			"por encima": (s > lim_sup).sum(),
			"total outliers": mask.sum(),
			"% outliers": mask.mean() * 100,
		}
	)


# En CO2 y arena importada se evalúan sólo los registros que usaron el insumo
df_out = df_dist.copy()
for col in ["co2_inyectado_m3", "arena_bombeada_importada_tn"]:
	df_out[col] = df_out[col].replace(0, np.nan)

tabla_outliers = df_out.apply(outliers_iqr).T.round(2)
display(tabla_outliers)
````

### 2.5.6 Análisis de correlación entre variables

Se analiza la relación entre las variables numéricas para identificar:

* Qué variables están asociadas a la variable objetivo (`agua_inyectada_m3`) y pueden ser buenas predictoras.
* Variables muy correlacionadas entre sí, que aportan información redundante y pueden generar problemas de multicolinealidad en el modelo.


Como la prueba de normalidad rechazó H₀ en todas las variables y se detectaron outliers y colas pesadas, se utiliza el **coeficiente de Spearman** (no paramétrico).

#### Prueba de hipótesis de correlación

Para cada variable se evalúa si su correlación con la variable objetivo es estadísticamente significativa:

***H0 : No existe correlación entre la variable estudiada y el agua inyectada (ρ = 0)***

***Ha : Existe correlación entre la variable estudiada y el agua inyectada (ρ ≠ 0)***

*Nivel de significancia: 𝛼 = 0.05*


````python
# @title Matriz de correlación de Spearman
corr_spearman = df_dist.corr(method="spearman")

# Se muestra sólo el triángulo inferior (la matriz es simétrica)
corr_tri = corr_spearman.iloc[1:, :-1]
mascara = np.triu(np.ones_like(corr_tri, dtype=bool), k=1)

with sns.axes_style("white"):
    fig, ax = plt.subplots(figsize=(11, 9))

sns.heatmap(corr_tri, mask=mascara, annot=True, fmt=".2f", cmap="RdBu_r",
            vmin=-1, vmax=1, center=0, square=True, linewidths=0.5,
            cbar_kws={"label": "Coeficiente de Spearman (ρ)", "shrink": 0.8}, ax=ax)
ax.tick_params(axis="x", rotation=45)
plt.setp(ax.get_xticklabels(), ha="right")

plt.suptitle(
    "Matriz de correlación de Spearman entre las variables numéricas\n"
    "Operaciones de fractura hidráulica en pozos de hidrocarburos en Argentina",
    fontsize=12, fontweight="bold",
)
fig.text(0.5, 0.005,
         "Fuente: elaboración propia con base en datos de la Secretaría de Energía de la Nación",
         ha="center", va="bottom", fontsize=8, style="italic")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.show()
````

````python
# @title Correlación de cada variable con la variable objetivo
filas = []
for col in num_cols:
    if col == "agua_inyectada_m3":
        continue
    pares = df_dist[["agua_inyectada_m3", col]].dropna()
    rho, p_valor = stats.spearmanr(pares["agua_inyectada_m3"], pares[col])
    filas.append({
        "variable": col,
        "n": len(pares),
        "rho de Spearman": round(rho, 3),
        "p-valor": p_valor,
        "decisión (α = 0,05)": "Se rechaza H0" if p_valor < 0.05 else "No se rechaza H0",
    })

correlacion_objetivo = (pd.DataFrame(filas)
                        .sort_values("rho de Spearman", key=abs, ascending=False)
                        .reset_index(drop=True))
display(correlacion_objetivo)
````

**Relación con la variable objetivo**
En todas las variables el p-value es menor que α = 0,05, por lo que se rechaza H₀; es decir que todas presentan una correlación estadísticamente significativa con el agua inyectada. Sin embargo, con casi 5.000 registros incluso correlaciones débiles resultan significativas, por lo que lo relevante es la magnitud de ρ:

Esto confirma lo planteado en el análisis de distribución, ya que se observa que las variables que describen el tamaño de la operación (longitud de la rama, cantidad de etapas y arena) son las más asociadas al volumen de agua.

**Redundancia entre variables (multicolinealidad):** la longitud de la rama horizontal, la cantidad de fracturas y la arena nacional están muy correlacionadas entre sí (ρ entre 0,89 y 0,93). Las tres miden, en esencia, el tamaño del pozo. Si se incluyen juntas en un modelo lineal pueden generar multicolinealidad, por lo que en la Fase 3 conviene evaluar quedarse con una o dos de ellas, o usar modelos que no se vean afectados por este problema.

**Efecto de la mezcla de poblaciones:** parte de la intensidad de estas correlaciones se explica porque el dataset combina pozos verticales (poca agua, sin rama, pocas etapas) con pozos horizontales (mucha agua, rama larga, muchas etapas). Esta diferencia entre grupos hace que todas las variables "crezcan juntas" y puede inflar los coeficientes. #### 2.5.7.2 Relación entre variables categóricas

Se analiza si las variables categóricas están asociadas entre sí, mediante tablas de contingencia y la **prueba chi-cuadrado de independencia**. Si dos variables están fuertemente asociadas, aportan información redundante al modelo.

Se analizan tres pares de variables:

* **Tipo de reservorio vs. cuenca:** para ver si cada cuenca se especializa en un tipo de reservorio.
* **Tipo de terminación vs. cuenca:** para ver si la técnica de terminación depende de la cuenca.
* **Subtipo de reservorio vs. tipo de terminación:** para ver si la técnica de terminación depende del tipo de roca.

##### Prueba de hipótesis: chi-cuadrado de independencia

***H0 : Las dos variables son independientes***

***Ha : Las dos variables no son independientes (Existe asociación entre ellas)***

*Nivel de significancia: 𝛼 = 0.05*

**Supuesto del test:** las frecuencias esperadas deben ser mayores o iguales a 5 en al menos el 80 % de las celdas. Por eso se excluye la categoría "NO DISCRIMINADO" de `tipo_reservorio`, que tiene un único registro.

**Intensidad de la asociación:** el test sólo indica si existe asociación, no cuán fuerte es. Para medir su intensidad se calcula la **V de Cramér**.

````python
# @title Tablas de contingencia
df_contingencia = df_datos_energia[df_datos_energia["tipo_reservorio"] != "NO DISCRIMINADO"]

pares = [("tipo_reservorio", "cuenca"),
         ("tipo_terminacion", "cuenca"),
         ("subtipo_reservorio", "tipo_terminacion")]

for v1, v2 in pares:
    print(f"\n{v1} vs. {v2}")
    display(pd.crosstab(df_contingencia[v1], df_contingencia[v2], margins=True, margins_name="Total"))
````

````python
# @title Prueba chi-cuadrado de independencia y V de Cramér
filas = []
for v1, v2 in pares:
    tabla = pd.crosstab(df_contingencia[v1], df_contingencia[v2])
    chi2, p_valor, gl, esperadas = stats.chi2_contingency(tabla)
    valor_critico = stats.chi2.ppf(1 - 0.05, gl)
    n = tabla.to_numpy().sum()
    v_cramer = np.sqrt(chi2 / (n * (min(tabla.shape) - 1)))

    filas.append({
        "variables": f"{v1} vs. {v2}",
        "n": n,
        "gl": gl,
        "χ²": round(chi2, 1),
        "valor crítico χ² (α = 0,05)": round(valor_critico, 2),
        "p-valor": "< 0,001" if p_valor < 0.001 else round(p_valor, 4),
        "decisión": "Se rechaza H0" if chi2 > valor_critico else "No se rechaza H0",
        "V de Cramér": round(v_cramer, 3),
        "% celdas con esperada < 5": round((esperadas < 5).mean() * 100, 1),
    })

chi2_resultados = pd.DataFrame(filas)
display(chi2_resultados)
````

En los tres pares de variables el estadístico χ² supera ampliamente el valor crítico, por lo que se rechaza la hipótesis nula, es decir que todas las variables analizadas están asociadas entre sí. En todos los casos se cumple el supuesto del test.

* **Tipo de reservorio vs. cuenca (V de Cramér = 0,74, asociación fuerte):** cada cuenca se especializa en un tipo de reservorio: el 96 % de los registros de Golfo San Jorge son convencionales, mientras que el 97 % de los de Austral y el 91 % de los de Neuquina son no convencionales.

* **Subtipo de reservorio vs. tipo de terminación (V de Cramér = 0,64, asociación fuerte):** la técnica de terminación depende del tipo de roca: el *shale* se completa principalmente con "Tapón disparo", mientras que el *tight* se asocia con otras técnicas.

* **Tipo de terminación vs. cuenca (V de Cramér = 0,35, asociación moderada):** "Tapón disparo", "Camisas y punzados" y "Jetteo" se usan casi exclusivamente en la cuenca Neuquina. En cambio, "Punzado" se usa en las tres cuencas, por eso la asociación es sólo moderada.

**Implicancias para el modelo:** las variables categóricas están fuertemente relacionadas entre sí, y en buena parte describen lo mismo: si la operación es un pozo no convencional de Vaca Muerta o no. Incluirlas todas en el modelo aportaría información redundante. Esto se tendrá en cuenta en la Fase 3 al seleccionar las variables predictoras.

#### 2.5.7.3 Relación con la variable objetivo

##### Prueba de hipótesis: comparación entre grupos

Para evaluar si las diferencias observadas entre grupos son estadísticamente significativas se utiliza la **prueba de Kruskal-Wallis**. Es la alternativa no paramétrica al ANOVA, ya que no requiere normalidad y compara las distribuciones de dos o más grupos a partir de los rangos de los datos.

***H0 : La distribución del agua inyectada es la misma en todos los grupos de la variable***

***Ha : Al menos un grupo presenta una distribución diferente del agua inyectada***

*Nivel de significancia: 𝛼 = 0.05*


````python
# @title Prueba de Kruskal-Wallis para agua inyectada según cada variable categórica
df_cat = df_datos_energia[df_datos_energia["agua_inyectada_m3"] > 0]

cat_cols = ["tipo_reservorio", "subtipo_reservorio", "tipo_terminacion",
            "formacion_productiva", "empresa_informante"]

filas = []
for col in cat_cols:
    medianas = df_cat.groupby(col)["agua_inyectada_m3"].agg(["count", "median"])
    medianas = medianas[medianas["count"] >= 5]
    grupos = [df_cat.loc[df_cat[col] == g, "agua_inyectada_m3"] for g in medianas.index]

    h, p_valor = stats.kruskal(*grupos)
    gl = len(grupos) - 1
    valor_critico = stats.chi2.ppf(1 - 0.05, gl)

    filas.append({
        "variable": col,
        "grupos (k)": len(grupos),
        "gl (k - 1)": gl,
        "H de Kruskal-Wallis": round(h, 1),
        "valor crítico χ² (α = 0,05)": round(valor_critico, 2),
        "p-valor": p_valor,
        "decisión": "Se rechaza H0" if h > valor_critico else "No se rechaza H0",
        "grupo con mayor mediana": f"{medianas['median'].idxmax()} ({medianas['median'].max():,.0f} m³)",
        "grupo con menor mediana": f"{medianas['median'].idxmin()} ({medianas['median'].min():,.0f} m³)",
    })

kruskal_resultados = pd.DataFrame(filas)
display(kruskal_resultados)
````

En las cinco variables el estadístico H supera ampliamente el valor crítico de la chi-cuadrado. En consecuencia, y en coincidencia con p-values menores a 0,001 en todos los casos, se rechaza H₀. Es decir que el agua inyectada difiere significativamente entre los grupos de todas las variables categóricas. Al igual que en las pruebas anteriores, con casi 5.000 registros el test detecta incluso diferencias pequeñas, por lo que lo relevante es la magnitud de las diferencias entre medianas, que en este caso es muy grande:

* **Tipo de reservorio:** los no convencionales tienen una mediana de 41.686 m³, frente a 104 m³ de los convencionales.

* **Subtipo de reservorio:** el *shale* tiene una mediana de 54.165 m³, unas 40 veces mayor que el *tight* (1.388 m³). Esto confirma lo observado en los diagramas de dispersión: los pozos *tight* usan mucha menos agua.

* **Tipo de terminación:** "Camisas y punzados" (79.623 m³) y "Tapón disparo" (52.199 m³) concentran los mayores volúmenes, mientras que "Punzado" tiene una mediana de 602 m³ y la mayor dispersión de todos los grupos.

* **Formación productiva:** Vaca Muerta tiene la mayor mediana (54.271 m³) y concentra la mayoría de los registros.

* **Empresa operadora:** las medianas van de 58 m³ a 80.767 m³ según la operadora.



**Las variables categóricas reflejan el tipo de pozo:** los grupos con más agua coinciden siempre con los pozos horizontales no convencionales de Vaca Muerta. Es decir, estas variables están asociadas al agua, pero en buena parte porque indican de qué tipo de operación se trata. Esto se debe tener en cuenta al seleccionar predictoras, porque pueden aportar información redundante con la longitud de la rama o la cantidad de etapas.

**Problemas de calidad detectados:**

* **Nombres de empresas duplicados:** una misma operadora aparece con distintos nombres (por ejemplo, "PLUSPETROL S.A." y "PLUSPETROL CUENCA NEUQUINA S.R.L.", o "VISTA ENERGY ARGENTINA SAU" y "VISTA OIL & GAS ARGENTINA SAU"). Deben unificarse en la Fase 3.
* **Categorías con muy pocos registros:** `formacion_productiva` tiene 28 categorías y `empresa_informante` 23, varias con menos de 5 registros, que se excluyeron de la prueba. En la Fase 3 conviene agruparlas en una categoría "Otras".
* **Formaciones mal cargadas:** aparecen valores como "formación improductiva" y "pozo d-129", que no son nombres de formaciones geológicas.

##### Prueba post-hoc: test de Dunn

La prueba de Kruskal-Wallis indica que existen diferencias entre los grupos, pero no cuáles grupos difieren entre sí. Para identificarlos se aplica el **test de Dunn**, que compara todos los pares de grupos a partir de los rangos de los datos.

**Test de Dunn**

***H0 : La distribución del agua inyectada es la misma en ambos grupos***

***Ha : La distribución del agua inyectada difiere entre ambos grupos***

*Nivel de significancia: 𝛼 = 0.05*

Como se realizan varias comparaciones a la vez, aumenta la probabilidad de encontrar alguna diferencia significativa sólo por azar. Para controlarlo se aplica la **corrección de Bonferroni**, que multiplica cada p-valor por la cantidad de comparaciones realizadas.

El test se aplica sólo al **tipo de terminación**, que tiene 5 grupos (10 comparaciones):

* El tipo y el subtipo de reservorio tienen sólo 2 grupos, por lo que la diferencia detectada por Kruskal-Wallis ya identifica qué grupos difieren.

* La formación productiva y la empresa tienen más de 20 grupos, muchos con pocos registros. El test se aplicará una vez agrupadas las categorías en la Fase 3.

````python
# @title Test de Dunn con corrección de Bonferroni: tipo de terminación
dunn_terminacion = sp.posthoc_dunn(df_cat, val_col="agua_inyectada_m3",
                                   group_col="tipo_terminacion", p_adjust="bonferroni")

display(dunn_terminacion.map(lambda p: "< 0,001" if p < 0.001 else round(p, 4)))
````

La matriz muestra el p-value ajustado por Bonferroni para cada par de tipos de terminación. De las 10 comparaciones, 7 presentan un p ajustado < 0,001, por lo que se rechaza H₀ y se concluye que esos grupos difieren en el agua inyectada. En las 3 comparaciones restantes el p ajustado es 1,0, por lo que no se rechaza H₀:

* "Camisas deslizables" vs. "Jetteo"
* "Camisas deslizables" vs. "Punzado"
* "Jetteo" vs. "Punzado"

Esto permite agrupar las técnicas de terminación en **tres grupos homogéneos** según el agua inyectada:

1. **Camisas y punzados** (mediana de 79.623 m³): el grupo de mayor consumo, que difiere de todos los demás.

2. **Tapón disparo** (mediana de 52.199 m³): también difiere de todos los demás.

3. **Camisas deslizables, Jetteo y Punzado** (medianas de 3.492, 1.374 y 602 m³): no se diferencian entre sí, por lo que se comportan como un único grupo de bajo consumo.

### 2.5.8 Evolución temporal del agua inyectada

Se analiza cómo varió el agua inyectada a lo largo del tiempo, a partir del año de inicio de la fractura (`anio_if`). Esta variable es cuantitativa discreta, por lo que su relación con la variable objetivo se evalúa con el coeficiente de Spearman, al igual que en el análisis de correlación.

Además, se describe la evolución por separado para los pozos horizontales *shale* de la cuenca Neuquina, que son el grupo de mayor consumo de agua, para distinguir si el aumento se debe a un cambio en el tipo de operaciones realizadas o a cambios dentro de un mismo tipo de pozo. Esta desagregación es sólo descriptiva y no modifica el dataset.


#### Prueba de hipótesis: correlación entre el año y el agua inyectada

***H0 : No existe correlación entre el año de inicio de la fractura y el agua inyectada (ρ = 0)***

***Ha : Existe correlación entre el año de inicio de la fractura y el agua inyectada (ρ ≠ 0)***

*Nivel de significancia: 𝛼 = 0.05*

````python
# @title Correlación de Spearman entre el año de inicio y el agua inyectada
df_temp = df_datos_energia[df_datos_energia["agua_inyectada_m3"] > 0].copy()

rho, p_valor = stats.spearmanr(df_temp["anio_if"], df_temp["agua_inyectada_m3"])
n = len(df_temp)
t = rho * np.sqrt((n - 2) / (1 - rho ** 2))
t_critico = stats.t.ppf(1 - 0.05 / 2, n - 2)

print(f"n = {n}")
print(f"ρ de Spearman = {rho:.3f}")
print(f"Estadístico t = {t:.2f}")
print(f"Valor crítico t (α = 0,05, bilateral, {n - 2} gl) = {t_critico:.2f}")
print(f"p-valor: {'< 0,001' if p_valor < 0.001 else round(p_valor, 4)}")
print(f"Decisión: {'Se rechaza H0' if abs(t) > t_critico else 'No se rechaza H0'}")
````

### 2.5.9. Verificación de las variables temporales derivadas

#### Análisis de variables temporales redundantes

Se verifica la integridad de las ocho variables temporales derivadas (`anio_if`, `mes_if`, `anio_ff`, `mes_ff`, `anio_carga`, `mes_carga`, `anio` y `mes`).


````python
# @title 2.5.9 Verificación de las variables temporales derivadas

# Columnas derivadas y sus fechas de origen.
# Verificar que mes/anio correspondan efectivamente a fecha_data.
columnas_derivadas = {
	"anio_if": ("fecha_inicio_fractura", "year"),
	"mes_if": ("fecha_inicio_fractura", "month"),
	"anio_ff": ("fecha_fin_fractura", "year"),
	"mes_ff": ("fecha_fin_fractura", "month"),
	"anio_carga": ("fecha_data", "year"),
	"mes_carga": ("fecha_data", "month"),
	"anio": ("fecha_inicio_fractura", "year"),
	"mes": ("fecha_inicio_fractura", "month"),
}

# Crear una copia temporal: no se modifica todavía el dataset.
df_revision = df_datos_energia.copy()

# Convertir temporalmente las fechas para poder compararlas.
fechas = {
	columna: pd.to_datetime(df_revision[columna], errors="coerce")
	for columna in [
		"fecha_inicio_fractura",
		"fecha_fin_fractura",
		"fecha_data",
	]
	if columna in df_revision.columns
}

resultados = []

for columna, (fecha_origen, componente) in columnas_derivadas.items():
	if columna not in df_revision.columns:
		resultados.append(
			{
				"columna": columna,
				"estado": "COLUMNA AUSENTE",
			}
		)
		continue

	if fecha_origen not in fechas:
		resultados.append(
			{
				"columna": columna,
				"estado": f"FALTA FECHA ORIGEN: {fecha_origen}",
			}
		)
		continue

	fecha = fechas[fecha_origen]
	esperado = getattr(fecha.dt, componente)
	observado = pd.to_numeric(df_revision[columna], errors="coerce")

	fecha_valida = fecha.notna()
	valor_presente = observado.notna()

	# Filas que pueden compararse: fecha válida y valor derivado presente.
	comparables = fecha_valida & valor_presente

	# Incluir valores derivados faltantes cuando la fecha sí es válida.
	discrepancias = fecha_valida & (observado.ne(esperado) | observado.isna())

	resultados.append(
		{
			"columna": columna,
			"fecha_origen": fecha_origen,
			"nulos_derivada": int(df_revision[columna].isna().sum()),
			"fechas_origen_invalidas_o_nulas": int((~fecha_valida).sum()),
			"filas_comparables": int(comparables.sum()),
			"discrepancias": int(discrepancias.sum()),
		}
	)

	if discrepancias.any():
		print(f"\nDiscrepancias en {columna}:")
		print(
			pd.DataFrame(
				{
					"fecha_origen": fecha[discrepancias],
					"valor_observado": observado[discrepancias],
					"valor_esperado": esperado[discrepancias],
				}
			).head(10)
		)

# Resumen general
df_integridad_fechas = pd.DataFrame(resultados)
display(df_integridad_fechas)
````

Las ocho variables temporales derivadas coinciden en el 100 % de los registros con el año o el mes de su fecha de origen. No se encontraron discrepancias ni valores faltantes. Por lo tanto, estas columnas no presentan errores de carga.

Sin embargo, son **redundantes**: todos sus datos pueden obtenerse directamente de `fecha_inicio_fractura`, `fecha_fin_fractura` y `fecha_data`. Además, `anio` y `mes` son copias exactas de `anio_if` y `mes_if`. En la Fase 3 se recomienda eliminarlas y, cuando sea necesario, recalcular los componentes temporales a partir de las fechas originales.

A diferencia de `potencia_equipos_fractura_hp`, la presión no define la población de estudio ni entra al modelo: describe un resultado de la operación. Por eso los valores imposibles se convierten en faltantes y **se conserva el registro**, que sigue siendo válido para el resto de las mediciones. Los casos extremos se revisan en la fase 5.

````python
# @title Marcar como faltantes los valores de presión sin sentido físico
presion = pd.to_numeric(df_limpio["presion_maxima_psi"], errors="coerce")
fuera_de_rango = presion.notna() & ~presion.between(1000, 30000)

print("Presión imposible (< 1.000 psi o > 30.000 psi):", int(fuera_de_rango.sum()))
print("  de ellos, registrados en cero:", int((presion == 0).sum()))
df_limpio.loc[fuera_de_rango, "presion_maxima_psi"] = np.nan
print("Presión nula tras el tratamiento:", int(df_limpio["presion_maxima_psi"].isna().sum()))
````

### 3.1.5. Valores imposibles en `potencia_equipos_fractura_hp`

Además de los ceros, la Fase 2 (2.5.4 y 2.5.5) identificó potencias físicamente imposibles: registros con más de 200.000 hp, unas cinco veces el valor típico de una operación (≈ 40.000 hp), y registros con menos de 100 hp, insuficientes para un equipo de fractura. Como la potencia no se utiliza como entrada del modelo (ver 3.5), estos valores se convierten en faltantes y se conserva el registro.

**Criterio de limpieza:** marcar como faltantes las potencias menores a 100 hp o mayores a 200.000 hp.

````python
# @title Marcar como faltantes los valores de potencia sin sentido físico
potencia = pd.to_numeric(df_limpio["potencia_equipos_fractura_hp"], errors="coerce")
potencia_imposible = (potencia < 100) | (potencia > 200_000)

print("Potencia imposible (< 100 hp o > 200.000 hp):", int(potencia_imposible.sum()))
df_limpio.loc[potencia_imposible, "potencia_equipos_fractura_hp"] = np.nan
````

### 3.1.6. Inconsistencias en las fechas de la operación

La Fase 2 (2.5.5) identificó operaciones con fecha de fin anterior a la de inicio y operaciones con una duración mayor a un año, que no corresponden a una única operación de fractura. Como no es posible determinar cuál de las dos fechas es la incorrecta, y la fecha de inicio se utiliza para delimitar la población de estudio (3.4), se conserva la fecha de inicio y se marca como faltante la fecha de fin. El registro se conserva, ya que la duración no se utiliza como entrada del modelo.

**Criterio de limpieza:** marcar como faltante `fecha_fin_fractura` cuando la duración de la operación es negativa o mayor a 365 días.

````python
# @title Marcar como faltante la fecha de fin inconsistente
inicio = pd.to_datetime(df_limpio["fecha_inicio_fractura"], errors="coerce")
fin = pd.to_datetime(df_limpio["fecha_fin_fractura"], errors="coerce")
duracion_dias = (fin - inicio).dt.days

fecha_fin_inconsistente = (duracion_dias < 0) | (duracion_dias > 365)
print("Duración negativa:", int((duracion_dias < 0).sum()))
print("Duración mayor a 365 días:", int((duracion_dias > 365).sum()))
df_limpio.loc[fecha_fin_inconsistente, "fecha_fin_fractura"] = pd.NaT
````

## 3.4. Selección de la población de estudio

La fase 1 define el análisis de reservorios **no convencionales de la cuenca Neuquina**. La fase 2 muestra diferencias fuertes de volumen entre cuencas, tipos de pozo y subtipos; mezclar esos grupos dificultaría interpretar el modelo. Se seleccionan operaciones con `cuenca = NEUQUINA`, `tipo_reservorio = NO CONVENCIONAL`, `subtipo_reservorio = SHALE` y `longitud_rama_horizontal_m > 0`. La longitud positiva es el criterio operativo de pozo horizontal: la fase 2 documenta que cero corresponde a un pozo vertical. Se conserva `formacion_productiva` sin imponer que sea Vaca Muerta. La cantidad de filas excluidas se calcula abajo, sin asumirla.

Esta decisión restringe las conclusiones al grupo seleccionado. El agua registrada en cero se identificó en la fase 2 como inválida para entrenar y se excluye aquí. Fechas posteriores a la revisión del 28/10/2026 se dejan fuera por no ser operaciones históricas observables a esa fecha.


````python
# @title Delimitar el dominio y excluir objetivos inválidos
fecha_revision = pd.Timestamp("2026-10-28")
filas_preparadas = len(df_limpio)
alcance = (
	df_limpio["cuenca"].eq("NEUQUINA")
	& df_limpio["tipo_reservorio"].eq("NO CONVENCIONAL")
	& df_limpio["subtipo_reservorio"].eq("SHALE")
	& df_limpio["longitud_rama_horizontal_m"].gt(0)
)
agua = pd.to_numeric(df_limpio["agua_inyectada_m3"], errors="coerce")
fecha_inicio = pd.to_datetime(df_limpio["fecha_inicio_fractura"], errors="coerce")
valido = alcance & agua.gt(0) & np.isfinite(agua) & fecha_inicio.notna() & fecha_inicio.le(fecha_revision)
print(f"Registros tras limpieza inicial: {filas_preparadas:,}")
print(f"Dentro del alcance geológico y operativo: {alcance.sum():,}")
print(f"Agua nula/no positiva dentro del alcance: {(alcance & ~agua.gt(0)).sum():,}")
print(f"Fechas posteriores a la revisión dentro del alcance: {(alcance & fecha_inicio.gt(fecha_revision)).sum():,}")
df_limpio = df_limpio.loc[valido].copy()
print(f"Registros finales para modelado: {len(df_limpio):,}")
print("Formaciones en el conjunto final:")
print(df_limpio["formacion_productiva"].value_counts())
````

### 3.4.3. Agrupación del tipo de terminación

En la Fase 2 el test de Dunn se aplicó sobre todo el dataset. Como la composición de los grupos cambia al acotar la población, se repite el test sobre el conjunto seleccionado antes de agrupar categorías.

***H0 : La distribución del agua inyectada es la misma en ambos grupos***

***Ha : La distribución del agua inyectada difiere entre ambos grupos***

*Nivel de significancia: 𝛼 = 0.05, con corrección de Bonferroni*

````python
# @title Test de Dunn sobre la población seleccionada
display(df_limpio["tipo_terminacion"].value_counts().to_frame("registros"))

dunn_poblacion = sp.posthoc_dunn(df_limpio, val_col="agua_inyectada_m3",
                                 group_col="tipo_terminacion", p_adjust="bonferroni")
display(dunn_poblacion.map(lambda p: "< 0,001" if p < 0.001 else round(p, 4)))
display(df_limpio.groupby("tipo_terminacion")["agua_inyectada_m3"].median().round(0)
                 .to_frame("mediana agua (m³)"))
````

Dentro de la población seleccionada, "Punzado" difiere de las demás técnicas (mediana de 51.021 m³), por lo que la agrupación en tres grupos observada en la Fase 2 sobre todo el dataset no se mantiene.

"Jetteo" tiene sólo 2 registros, por lo que no puede conservarse como categoría propia. Con tan pocos casos el test tiene poca potencia y no detecta diferencias entre "Jetteo" y "Punzado" ni entre "Jetteo" y "Tapón disparo"; la comparación más clara es con "Camisas deslizables", con un p ajustado de 1,0. Por eso se unen "Jetteo" y "Camisas deslizables" en una única categoría, y la variable queda con 4 categorías.

````python
# @title Agrupación de Camisas deslizables y Jetteo
df_limpio["tipo_terminacion"] = df_limpio["tipo_terminacion"].replace({
    "Camisas deslizables": "Camisas deslizables / Jetteo",
    "Jetteo": "Camisas deslizables / Jetteo",
})
display(df_limpio["tipo_terminacion"].value_counts().to_frame("registros"))
````

La fase 4 lee `df_limpio.csv`, generado por la fase 3 desde el CSV original versionado. El archivo ya contiene el recorte a pozos horizontales shale de la cuenca Neuquina y la exclusión de objetivos inválidos y fechas futuras al 28/10/2026. Se comprueban estas condiciones para detectar una salida vieja o equivocada. La selección de la población y la limpieza se documentan en la fase 3; aquí solo se preparan las columnas de entrada para el algoritmo.


````python
# @title Carga del conjunto preparado en la Fase 3
if not RUTA_DATOS_LIMPIOS.is_file():
	raise FileNotFoundError("Ejecutá primero la fase 3 desde la raíz del repositorio.")
df = pd.read_csv(RUTA_DATOS_LIMPIOS)
df["fecha_inicio_fractura"] = pd.to_datetime(df["fecha_inicio_fractura"], errors="coerce")
print(f"Registros recibidos de fase 3: {len(df):,}")
````

````python
# La lista de entradas y el objetivo viven en comun.py para que la fase 5
# evalúe exactamente las mismas variables con las que se entrenó el modelo.
objetivo = OBJETIVO
entradas_numericas = ENTRADAS_NUMERICAS
entradas_categoricas = ENTRADAS_CATEGORICAS
entradas = entradas_numericas + entradas_categoricas
columnas_necesarias = [*entradas, objetivo, "cuenca", "tipo_reservorio", "subtipo_reservorio", "idpozo", "id_base_fractura_adjiv", "fecha_inicio_fractura"]
faltantes = set(columnas_necesarias) - set(df.columns)
if faltantes:
	raise ValueError(f"Columnas faltantes en la salida de fase 3: {sorted(faltantes)}")

alcance_correcto = (
	df["cuenca"].eq("NEUQUINA") & df["tipo_reservorio"].eq("NO CONVENCIONAL") & df["subtipo_reservorio"].eq("SHALE") & df["longitud_rama_horizontal_m"].gt(0)
)
agua = pd.to_numeric(df[objetivo], errors="coerce")
if not alcance_correcto.all() or not (agua.gt(0) & np.isfinite(agua)).all() or not df["fecha_inicio_fractura"].le(FECHA_REVISION).all():
	raise ValueError("La salida de fase 3 no cumple el alcance o los criterios de calidad documentados; ejecutá de nuevo esa fase.")

datos_modelo = df[columnas_necesarias].copy()
datos_modelo[objetivo] = agua
for columna in entradas_numericas:
	datos_modelo[columna] = pd.to_numeric(datos_modelo[columna], errors="coerce").replace([np.inf, -np.inf], np.nan)
print(f"Registros listos para el modelo: {len(datos_modelo):,}")
print(datos_modelo[[*entradas, objetivo]].isna().sum().to_frame("faltantes antes del pipeline"))
````

## 4.2. Separación temporal

El corte de 2025 deja las operaciones posteriores exclusivamente para prueba. Si un pozo figura en ambos períodos, se retira del entrenamiento para evitar que el modelo aprenda información del mismo pozo que luego se evalúa. El conjunto de prueba no participa de la imputación, codificación ni elección de parámetros.


````python
corte_temporal = CORTE_TEMPORAL
train = datos_modelo.loc[datos_modelo["fecha_inicio_fractura"] < corte_temporal].copy()
test = datos_modelo.loc[datos_modelo["fecha_inicio_fractura"] >= corte_temporal].copy()
pozos_compartidos = set(train["idpozo"]).intersection(test["idpozo"])
train = train.loc[~train["idpozo"].isin(pozos_compartidos)].copy()
if train.empty or test.empty:
	raise ValueError("La partición temporal quedó vacía; revisar fechas y filtros.")
assert train["fecha_inicio_fractura"].max() < test["fecha_inicio_fractura"].min()
assert not set(train["idpozo"]).intersection(test["idpozo"])
X_train, y_train = train[entradas], train[objetivo]
X_test, y_test = test[entradas], test[objetivo]
print(f"Entrenamiento: {len(train):,} operaciones; prueba: {len(test):,}.")
print(f"Pozos compartidos retirados del entrenamiento: {len(pozos_compartidos)}.")
print(f"Rango de entrenamiento: {train.fecha_inicio_fractura.min().date()} a {train.fecha_inicio_fractura.max().date()}")
print(f"Rango de prueba: {test.fecha_inicio_fractura.min().date()} a {test.fecha_inicio_fractura.max().date()}")
````

## 4.3. Entrenamiento

Las medianas numéricas y las categorías más frecuentes se calculan solo sobre entrenamiento. `OneHotEncoder` transforma las categorías y tolera categorías nuevas en prueba. Se fijan semilla y parámetros para obtener resultados reproducibles. `min_samples_leaf=3` limita hojas demasiado específicas; 200 árboles reducen la variación entre árboles con un costo razonable para este conjunto. Son parámetros iniciales, no se afirma que sean óptimos: su ajuste requeriría una validación temporal adicional dentro del período de entrenamiento.


````python
try:
	codificador = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:  # Compatibilidad con scikit-learn anterior a 1.2
	codificador = OneHotEncoder(handle_unknown="ignore", sparse=False)

prep = ColumnTransformer(
	[
		("num", SimpleImputer(strategy="median"), entradas_numericas),
		(
			"cat",
			Pipeline(
				[
					("imputer", SimpleImputer(strategy="most_frequent")),
					("onehot", codificador),
				]
			),
			entradas_categoricas,
		),
	]
)
modelo = Pipeline(
	[
		("prep", prep),
		("regresor", RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=42, n_jobs=-1)),
	]
)
referencia = DummyRegressor(strategy="median")
modelo.fit(X_train, y_train)
referencia.fit(X_train, y_train)
print("Modelos entrenados. La prueba temporal permanece reservada para la fase 5.")
````

## 4.4. Comparación de modelos

La elección del algoritmo se resuelve con validación cruzada **sobre el conjunto de entrenamiento** (5 cortes cronológicos, ordenados por `fecha_inicio_fractura`), y no sobre la prueba temporal: usar la prueba para elegir el modelo haría que la evaluación final de la fase 5 quedara inflada. Se comparan:

- **Ridge** (regresión lineal regularizada): referencia lineal interpretable y el único que puede extrapolar fuera del rango histórico; necesita escalado.
- **SVR con núcleo RBF**: captura no linealidades; se escala y se entrena sobre `log1p(agua)` por la asimetría del objetivo.
- **Random Forest**: el modelo de la sección 4.3.
- **Gradient Boosting** (implementación por histogramas de scikit-learn): suele ser el más fuerte en datos tabulares.

Todos comparten el mismo preprocesamiento (imputación + one-hot); Ridge y SVR agregan escalado numérico dentro del pipeline, para que no haya fuga de información desde la validación. El ganador se reentrena con todo el entrenamiento y es el que se exporta a la fase 5. La métrica de selección es el **MAE en m³**, la principal según la fase 5, por su lectura directa en las unidades del problema.

**Criterio de desempate:** la validación cruzada tiene variabilidad entre cortes, por lo que una diferencia de MAE menor que esa variabilidad no permite afirmar que un modelo sea mejor que otro. Si la diferencia entre el MAE de un modelo y el del mejor es menor o igual que el desvío de la validación del mejor, ambos se consideran empatados. Entre los modelos empatados se prioriza Ridge, por ser el único lineal: es el más fácil de interpretar y el único que puede extrapolar, algo relevante porque la Fase 2 (2.5.8) mostró que el tamaño de los pozos y el agua inyectada crecen con el tiempo, y los modelos basados en árboles no pueden estimar valores mayores a los observados en el entrenamiento. Si Ridge no está entre los empatados, se elige el modelo de menor MAE.

````python
# @title Comparación de modelos con validación cruzada temporal
def construir(regresor, escalar=False, log_objetivo=False):
	"""Arma el mismo preprocesamiento para todos, con extras opcionales."""
	pasos_num = [("imputer", SimpleImputer(strategy="median"))]
	if escalar:
		pasos_num.append(("scaler", StandardScaler()))

	pipeline = Pipeline(
		[
			(
				"prep",
				ColumnTransformer(
					[
						("num", Pipeline(pasos_num), entradas_numericas),
						(
							"cat",
							Pipeline(
								[
									("imputer", SimpleImputer(strategy="most_frequent")),
									("onehot", codificador),
								]
							),
							entradas_categoricas,
						),
					]
				),
			),
			("regresor", regresor),
		]
	)
	if log_objetivo:
		return TransformedTargetRegressor(regressor=pipeline, func=np.log1p, inverse_func=np.expm1)
	return pipeline


candidatos = {
	"Ridge": construir(Ridge(alpha=1.0), escalar=True),
	"SVR (RBF)": construir(SVR(C=100.0, gamma="scale", epsilon=0.05), escalar=True, log_objetivo=True),
	"Random Forest": construir(RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=42, n_jobs=-1)),
	"Gradient Boosting": construir(HistGradientBoostingRegressor(max_iter=300, learning_rate=0.08, min_samples_leaf=20, random_state=42)),
}

# La validación cruzada es cronológica: cada corte entrena con lo anterior y
# valida con lo posterior, igual que el uso real del modelo.
orden = train.sort_values("fecha_inicio_fractura").index
X_cv, y_cv = X_train.loc[orden], y_train.loc[orden]
cv = TimeSeriesSplit(n_splits=5)

filas = []
for nombre, candidato in candidatos.items():
	mae = -cross_val_score(candidato, X_cv, y_cv, cv=cv, scoring="neg_mean_absolute_error")
	filas.append({"modelo": nombre, "MAE_cv_m3": mae.mean(), "desvio_cv_m3": mae.std()})

comparativa = pd.DataFrame(filas).sort_values("MAE_cv_m3").reset_index(drop=True)

# Criterio de desempate: los modelos cuya diferencia de MAE con el mejor es menor o igual
# que el desvío de la validación del mejor se consideran empatados; entre ellos se prioriza Ridge.
mejor = comparativa.loc[0]
comparativa["diferencia_con_el_mejor_m3"] = comparativa["MAE_cv_m3"] - mejor["MAE_cv_m3"]
comparativa["empatado_con_el_mejor"] = comparativa["diferencia_con_el_mejor_m3"] <= mejor["desvio_cv_m3"]
display(comparativa.round(1))

empatados = comparativa.loc[comparativa["empatado_con_el_mejor"], "modelo"].tolist()
nombre_modelo = "Ridge" if "Ridge" in empatados else mejor["modelo"]
print(f"Modelos empatados con el mejor ({mejor['modelo']}): {empatados}")

# El ganador se reentrena con todo el entrenamiento y es el que pasa a fase 5.
modelo = candidatos[nombre_modelo]
modelo.fit(X_train, y_train)
print(f"Modelo elegido: {nombre_modelo}")
````

Se guardan el pipeline completo ganador de la comparación, su nombre, la tabla comparativa, la referencia, la partición de prueba y los identificadores de cada operación. La fase 5 consume exactamente estos objetos: así evalúa el mismo modelo y las mismas filas con las que se entrenó.

````python
exportar_modelos(
	modelo=modelo,
	referencia=referencia,
	nombre_modelo=nombre_modelo,
	comparativa_modelos=comparativa,
	X_test=X_test,
	y_test=y_test,
	test_metadata=test[["id_base_fractura_adjiv", "idpozo", "fecha_inicio_fractura"]],
)
print(f"Mejor modelo exportado: {nombre_modelo}")
print(f"Artefactos guardados en {RUTA_ARTEFACTOS}")
````

````python
def calcular_metricas(nombre, real, pred):
	return {
		"modelo": nombre,
		"MAE_m3": mean_absolute_error(real, pred),
		"RMSE_m3": np.sqrt(mean_squared_error(real, pred)),
		"MedAE_m3": median_absolute_error(real, pred),
		"R2": r2_score(real, pred),
	}


metricas = pd.DataFrame(
	[
		calcular_metricas("Referencia: mediana", y_test, pred_referencia),
		calcular_metricas(nombre_modelo, y_test, pred_modelo),
	]
)
display(metricas.round(3))

mae_ref = metricas.loc[metricas["modelo"].eq("Referencia: mediana"), "MAE_m3"].iloc[0]
mae_modelo = metricas.loc[metricas["modelo"].eq(nombre_modelo), "MAE_m3"].iloc[0]
mejora = 100 * (mae_ref - mae_modelo) / mae_ref
decision = "mejora la referencia" if mejora > 0 else "no mejora la referencia"
print(f"Decisión preliminar: el {nombre_modelo} {decision} en MAE ({mejora:.1f}%).")

# Los identificadores salen de test_metadata, guardado junto al resto de los
# artefactos: así cada fila coincide con su observación y su predicción.
resultados_test = test_metadata.reset_index(drop=True).copy()
resultados_test["agua_observada_m3"] = y_test.to_numpy()
resultados_test["agua_estimada_m3"] = pred_modelo
resultados_test["error_m3"] = resultados_test["agua_estimada_m3"] - resultados_test["agua_observada_m3"]
resultados_test["error_absoluto_m3"] = resultados_test["error_m3"].abs()
mayores_errores = resultados_test.nlargest(10, "error_absoluto_m3").copy()

columnas_numericas = mayores_errores.select_dtypes(include="number").columns
mayores_errores[columnas_numericas] = mayores_errores[columnas_numericas].round(1)

display(mayores_errores)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].scatter(y_test, pred_modelo, alpha=0.5, s=25)
limite = max(float(y_test.max()), float(pred_modelo.max()))
axes[0].plot([0, limite], [0, limite], "--", color="black", linewidth=1)
axes[0].set(title="Observado frente a estimado", xlabel="Agua observada (m³)", ylabel="Agua estimada (m³)")

sns.histplot(resultados_test["error_m3"], bins=35, ax=axes[1])
axes[1].axvline(0, color="black", linestyle="--", linewidth=1)
axes[1].set(title="Distribución del error", xlabel="Error: estimado - observado (m³)", ylabel="Registros")
plt.tight_layout()
plt.show()

# Importancia de variables según el tipo de modelo elegido.
if hasattr(modelo, "named_steps"):
	nucleo, prep = modelo.named_steps["regresor"], modelo.named_steps["prep"]
else:  # los modelos con objetivo transformado envuelven el pipeline
	nucleo, prep = modelo.regressor_.named_steps["regresor"], modelo.regressor_.named_steps["prep"]
nombres = prep.get_feature_names_out()

if hasattr(nucleo, "feature_importances_"):
	importancias = pd.Series(nucleo.feature_importances_, index=nombres, name="importancia").sort_values(ascending=False)
	display(importancias.head(12).to_frame().round(3))
elif hasattr(nucleo, "coef_"):
	coeficientes = pd.Series(np.abs(np.ravel(nucleo.coef_)), index=nombres, name="|coeficiente|").sort_values(ascending=False)
	display(coeficientes.head(12).to_frame().round(4))
else:
	resultado = permutation_importance(modelo, X_test, y_test, n_repeats=10, random_state=42, scoring="neg_mean_absolute_error")
	importancias = pd.Series(resultado.importances_mean, index=entradas, name="importancia por permutación").sort_values(ascending=False)
	display(importancias.to_frame().round(4))

print("Las importancias describen el uso de variables por el modelo; no prueban causalidad.")
````

````python
# @title Análisis de errores por año

resultados_anuales = resultados_test.copy()

resultados_anuales["anio"] = pd.to_datetime(
    resultados_anuales["fecha_inicio_fractura"]
).dt.year

resumen_anual = (
    resultados_anuales.groupby("anio")
    .agg(
        operaciones=("error_absoluto_m3", "size"),
        MAE_m3=("error_absoluto_m3", "mean"),
        MedAE_m3=("error_absoluto_m3", "median"),
    )
    .reset_index()
)

display(resumen_anual.round(2))
````

Consideramos esos tres modelos empatados y, según el criterio de desempate, elegimos Ridge por ser el único lineal: es más fácil de interpretar y puede seguir la tendencia de crecimiento de los pozos fuera del rango observado en el entrenamiento, mientras que los modelos basados en árboles no pueden estimar valores mayores a los del entrenamiento.

Estas operaciones forman una lista específica para revisar en detalle, ya que algunas podrían corresponder a errores de carga no detectados.

**Gráfico observado frente a estimado:** la línea punteada representa una estimación perfecta. Los puntos se distribuyen alrededor de esa diagonal, sin el techo en las estimaciones que presentan los modelos basados en árboles. La dispersión es algo mayor en los volúmenes altos, que es el principal límite del modelo.

**Distribución del error:** el error se concentra alrededor de cero, levemente desplazado hacia valores negativos, es decir que el modelo tiende a subestimar un poco el agua inyectada. Presenta colas hacia ambos lados por los desvíos mayores. Un error positivo significa que el modelo estimó más agua que la observada, y uno negativo, que la subestimó.

**Coeficientes:** sobre las variables estandarizadas, `cantidad_fracturas` es el término de mayor peso en valor absoluto (18.229,5), seguido por las categorías de `tipo_terminacion` (entre 990,0 y 17.726,2) y por `longitud_rama_horizontal_m` (2.247,6). Al tratarse de un modelo lineal con codificación one-hot completa, los coeficientes de las categorías se reparten con el intercepto y no deben interpretarse como efectos causales aislados: describen cómo se ajustó el modelo en este conjunto.

**Errores por año:** el MAE es de 8.107,7 m³ en 2025 y de 7.614,0 m³ en 2026, por lo que el desempeño se mantiene estable en los dos años del período de prueba. Hay que tener en cuenta que 2026 incluye sólo las operaciones disponibles hasta la fecha de revisión.

**Conclusión de la evaluación:** el modelo supera con claridad a la referencia, por lo que consideramos que permite avanzar hacia el despliegue. Su forma lineal lo hace más fácil de explicar y le permite extrapolar, lo que es útil para la planificación, pero también nos obliga a ser prudentes: una estimación alejada de los antecedentes históricos es poco confiable, y por eso la aplicación la señala como tal. En una próxima iteración proponemos revisar los valores extremos, analizar los errores por rango de volumen y comparar cada estimación con operaciones semejantes.

# Fase 6: Despliegue / Deployment
Por lo general, la creación del modelo no es el final del proyecto. Incluso si el propósito del modelo es aumentar el conocimiento de los datos, el conocimiento adquirido deberá organizarse y presentarse de manera que el cliente pueda utilizarlo. A menudo implica la aplicación de modelos "en vivo" dentro de los procesos de toma de decisiones de una organización, por ejemplo, en la personalización en tiempo real de páginas web o la puntuación repetida de bases de datos de marketing. Sin embargo, en función de los requisitos, la fase de implementación puede ser tan sencilla como generar un informe o tan compleja como implementar un proceso de minería de datos repetible en toda la empresa. En muchos casos, es el cliente, no el analista de datos, quien lleva a cabo los pasos de implementación. Sin embargo, incluso si el analista no llevará a cabo el esfuerzo de implementación, es importante que el cliente comprenda de antemano qué acciones deben llevarse a cabo para hacer uso de los modelos creados.
