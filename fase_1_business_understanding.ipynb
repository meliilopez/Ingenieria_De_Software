## 1.1. Problema de negocio

La planificación de las operaciones de fractura hidráulica requiere considerar distintos recursos, entre ellos el agua utilizada durante el proceso. Los registros públicos de operaciones realizadas en la cuenca Neuquina permiten estudiar cómo varía el volumen de agua inyectada según las características de los pozos y las operaciones.

El proyecto propone explorar estos registros para identificar relaciones entre el volumen de agua y variables como la longitud horizontal, la cantidad de fracturas, las características del reservorio, el tipo de terminación y los materiales utilizados. También se analizarán posibles diferencias entre períodos y grupos de operaciones.

A partir de los resultados, se evaluará si la información disponible permite estimar el volumen de agua de una operación o clasificar las operaciones según su consumo. La viabilidad de estas alternativas dependerá de la calidad, la cobertura y las limitaciones de los datos.

La solución se orienta a comunicar hallazgos y ofrecer referencias basadas en antecedentes históricos, sin reemplazar los cálculos de ingeniería ni la evaluación especializada.

**Fuente de datos:** [Datos de fractura de pozos — Adjunto IV de la Secretaría de Energía](https://datos.gob.ar/dataset/datos-de-fractura-de-pozos-adjunto-iv).

**Pregunta de estudio:** ¿En qué medida las características de los pozos y las operaciones permiten explicar o estimar el volumen de agua inyectada, y qué utilidad podrían tener estos resultados como referencia?

## 1.2. Objetivos del análisis

### Objetivo general

Analizar los registros históricos de fracturas hidráulicas en reservorios no convencionales de la cuenca Neuquina para estudiar la relación entre las características de los pozos y las operaciones y el volumen de agua inyectada, y evaluar cómo comunicar los resultados mediante una solución de software.

### Objetivos específicos

1. Caracterizar los datos disponibles e identificar sus principales limitaciones de cobertura, calidad y consistencia.
2. Explorar la distribución del volumen de agua inyectada y su relación con las características de las operaciones, los materiales utilizados y las variables temporales disponibles.
3. Investigar si los datos permiten estimar el volumen de agua inyectada o clasificar las operaciones según su consumo.
4. Evaluar las alternativas que resulten viables mediante métricas adecuadas y analizar sus limitaciones.
5. Comunicar los principales hallazgos mediante una solución de software sencilla y reproducible.

El trabajo seguirá las fases de CRISP-DM, adaptando las decisiones de preparación, análisis, modelado y evaluación a los resultados obtenidos.
